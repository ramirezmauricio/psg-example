# Introducción a Listas en Python
Ejemplos prácticos para entender la creación, acceso, modificación, slicing, métodos y operaciones con listas en Python.

## Lista de enteros

```python
print ("Lista de enteros")
mi_lista = [1,2,3,4,5]
print (mi_lista)
```

## Lista de cadenas

```python
print ("Lista de cadenas")
mi_lista = ["hola", "mundo", "python"]
print (mi_lista)
```

## Lista mixta

```python
print ("Lista mixta")
mi_lista = [1, "hola", 3.14, "mundo", 5]
print (mi_lista)
```

## Lista vacía

```python
print ("Lista vacía")
mi_lista = []
print (mi_lista)
```

## Lista a partir de una cadena

```python
print ("Lista a partir de una cadena")
mi_lista = list("hola mundo")
print (mi_lista)
```

## Lista a partir de una tupla

```python
print ("Lista a partir de una tupla")
mi_tupla = (1,2,3,4,5)
print (mi_tupla)
mi_lista = list(mi_tupla)
print (mi_lista)
```

## Lista por comprensión

```python
print ("Lista por comprensión")
mi_lista = [x for x in range(10)]
print (mi_lista)
```

## Acceso utilizando índices positivos

```python
print ("Indexación positivo de una lista")
lista = [1, "hola", 3.14, (1,2)]
print (lista[0], type(lista[0])) 
print (lista[1], type(lista[1])) 
print (lista[2], type(lista[2])) 
print (lista[3], type(lista[3])) 
```

## Acceso utilizando índices negativos

```python
print ("Indexación negativo de una lista")
lista = [1, "hola", 3.14, (1,2)]
print (lista[-1], type(lista[-1]))
print (lista[-2], type(lista[-2]))
print (lista[-3], type(lista[-3]))
print (lista[-4], type(lista[-4]))
```

## Modificación de una lista

La indexación no se limita a obtener valores, también permite modificar elementos de la lista.

```python
print ("Modificación de una lista")
lista = [1, "hola", 3.14, (1,2)]
print (lista)
lista[0] = 2
lista[1] = "mundo"
print (lista)
```

## Slicing de una lista

```python
print ("Slicing de una lista")
lista = ["P", "y", "t", "h", "o", "n", "L", "a", "P", "a", "z"]
print (lista)
sub_lista = lista[2:7]
print (sub_lista)
print (type(sub_lista))
```

## Concatenación de listas

```python
print ("Concatenación de listas")
lista1 = [1,2,3]
lista2 = ["a","b","c"]
concatenar = lista1 + lista2
print (lista1, lista2)
print (concatenar)
```

## Repetición de listas

```python
print ("Repetición de listas")
lista = [True, False]
repetir = lista * 3
print (repetir)
```

## Métodos de listas

### index(valor)

```python
print ("Método index(valor)")
lista = [1,True,3.14,"hola",5]
print (lista.index("hola"))
```

### count(valor)

```python
print ("Método count(valor)")
lista = [1,True,3.14,"hola",5, True, True, 3.140]
print (lista.count(True))
```

### insert(i, valor)

```python
print ("Método insert(i, valor)")
lista = [1,2,3,4,5]
lista.insert(2, "OwO")
print (lista)
```

## Copiar listas: referencia vs copia real

```python
print ("Asignación de lista")
lista = [1,2,3]
copia = lista
copia[0] = 9
print (lista)
print (copia)
```

### Copiar usando slicing [:] o copy()

```python
lista = [1,2,3]
copia = lista[:]
copia[0] = 9
print (lista)
print (copia)
```

## Funciones útiles para listas

```python
print ("Función len()")
lista = [1,2,3]
print (len(lista))
```

## Operadores de comparación y pertenencia

```python
print ("Comparación de listas")
lista = [1,2,3]
print (3 in lista)
print (6 not in lista)
```

## Listas anidadas

```python
print ("Listas anidadas")
lista = [1,2,3,[4,5,6]]
print (lista[3][1])
```
