# Diccionario de clave entera y valor entero

```python
print ("Diccionario de clave entera y valor entero")
diccionario = {1: 10, 2: 20, 3: 30}
print(diccionario)
print(type(diccionario))
```

# Diccionario de clave entero y valor cadena

```python
print ("Diccionario de clave entero y valor cadena")
diccionario = {1: 'uno', 2: 'dos', 3: 'tres'}
print(diccionario)
print(type(diccionario))
```

# Diccionario de clave cadena y valor entero

```python
print ("Diccionario de clave cadena y valor entero")
diccionario = {'uno': 1, 'dos': 2, 'tres': 3}
print(diccionario)
print(type(diccionario))
```

# Diccionario de clave cadena y valor cadena

```python
print ("Diccionario de clave cadena y valor cadena")
diccionario = {'perro': '🐶', 'gato': '🐱', 'ave': '🐦'}
print(diccionario)
print(type(diccionario))
```

# Diccionario mixto

```python
print ("Diccionario mixto")
diccionario = {
    1: "ID-XXXXX",
    "edad": 3,
    "animal": "🐶",
    ("John", "Doe"): 6917222722,
    "salvaje": False
}
print(diccionario)
print(type(diccionario))
```

# Crear diccionario vacío

```python
print ("Diccionario vacío")
diccionario = {}
print(diccionario)
print(type(diccionario))

diccionario = dict()
print(diccionario)
print(type(diccionario))
```

# Diccionario a partir de lista anidada

```python
print ("Diccionario a partir de una lista")
lista = [['perro', '🐶'], ['gato', '🐱'], ['ave', '🐦']]
print(lista)
diccionario = dict(lista)
print(diccionario)
print(type(diccionario))
```

# Diccionario a partir de tupla de tuplas

```python
print ("Diccionario a partir de una tupla de tuplas")
tupla = (('perro', '🐶'), ('gatos', '🐱'), ('ave', ['🐦', '🦅']))
print(tupla)
diccionario = dict(tupla)
print(diccionario)
print(type(diccionario))
```

# Diccionario mediante comprensión

```python
print ("Diccionario mediante comprensión")
diccionario = {animal: animal*3 for animal in '🐶🐱🐹🐰'}
print(diccionario)
print(type(diccionario))
```

 # Acceder mediante clave

```python
print ("Acceder mediante clave")
diccionario = {'perro': '🐶', 'gato': '🐱', 'ave': '🐦'}
print(diccionario)
print(diccionario['perro'], type(diccionario['perro']))
print(diccionario['gato'], type(diccionario['gato']))
print(diccionario['ave'], type(diccionario['ave']))
```

# Cambiar el valor de una clave

```python
print ("Cambiar el valor de una clave")
diccionario = {'perro': '🐶', 'gato': '🐱', 'ave': '🐦'}
diccionario['perro'] = '🐩'
print(diccionario)
```

# Crear nuevo par clave-valor

```python
print ("Crear un nuevo par clave-valor")
diccionario = {'perro': '🐶', 'gato': '🐱', 'ave': '🐦'}
diccionario['pez'] = '🐠'
print(diccionario)
```

# Eliminar un par clave-valor

```python
print ("Eliminar un par clave-valor")
diccionario = {'perro': '🐶', 'gato': '🐱', 'ave': '🐦'}
del diccionario['ave']
print(diccionario)
```

# Modificar una clave (crear nueva y eliminar antigua)

```python
print ("Modificar la clave")
diccionario = {'perro': '🐶', 'gato': '🐱', 'ave': '🐦'}
diccionario['perrito'] = diccionario['perro']
del diccionario['perro']
print(diccionario)
```

# Métodos de acceso: get()

```python
print ("Método get(clave)")
diccionario = {'perro': '🐶', 'gato': '🐱'}
print(diccionario.get('perro'))
print(diccionario.get('pez'))
```

# Métodos de acceso: items()

```python
print ("Método items()")
diccionario = {'perro': '🐶', 'gato': '🐱'}
print(list(diccionario.items()))
```

 # Métodos de acceso: keys() y values()

```python
print ("Método keys()")
diccionario = {'perro': '🐶', 'gato': '🐱'}
print(list(diccionario.keys()))

print ("Método values()")
print(list(diccionario.values()))
```

# Métodos de adición: update()

```python
print ("Método update(diccionario)")
diccionario = {'perro': '🐶', 'gato': '🐱'}
diccionario.update({'ave': '🐦'})
print(diccionario)

print ("Método update(clave=valor)")
diccionario.update(pez='🐠')
print(diccionario)
```

# Métodos de eliminación: clear(), pop(), popitem()

```python
print ("Método clear()")
diccionario = {'perro': '🐶', 'gato': '🐱'}
diccionario.clear()
print(diccionario)

print ("Método pop(clave)")
diccionario = {'perro': '🐶', 'gato': '🐱'}
diccionario.pop('gato')
print(diccionario)

print ("Método popitem()")
diccionario = {'perro': '🐶', 'gato': '🐱'}
print(diccionario.popitem())
print(diccionario)
```

# Copiar diccionario: referencia vs copy()

```python
print ("Asignación por referencia")
diccionario = {'perro': '🐶', 'gato': '🐱'}
copia = diccionario
copia['gato'] = '🦁'
print(diccionario)

print ("Método copy()")
diccionario = {'perro': '🐶', 'gato': '🐱'}
copia = diccionario.copy()
copia['gato'] = '🦁'
print(diccionario)
print(copia)
```

# Funciones: len(), in/not in, iter()

```python
print ("Función len()")
diccionario = {'perro': '🐶', 'gato': '🐱'}
print(len(diccionario))

print ("Función in / not in")
print('perro' in diccionario)
print('pez' not in diccionario)

print ("Función iter()")
iterador = iter(diccionario.items())
print(next(iterador))
```
