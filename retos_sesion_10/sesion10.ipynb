# Sesión 10: Conjuntos en Python

En esta sesión exploraremos cómo trabajar con conjuntos (`set`) en Python:  
cómo crearlos, usarlos, sus métodos, operaciones y ejemplos prácticos.

```python
# Conjunto de enteros
print ("Conjunto de enteros")
conjunto = {1, 2, 3, 4, 5}
print(conjunto) 
print(type(conjunto))
```

## Crear conjuntos con `set()`

Se utiliza la función `set()` para crear conjuntos vacíos o a partir de una secuencia.
Permite convertir listas, tuplas o cadenas en conjuntos.


```python
# Conjunto vacío
print ("Conjunto vacío")
conjunto = set()
print(conjunto)
print(type(conjunto))
```

```python
# Conjunto a partir de la cadena
print ("Conjunto a partir de la cadena")
cadena = 'Hola Mundo'
conjunto = set(cadena)
print(conjunto)
print(type(conjunto))
```

```python
# Conjunto a partir de una tupla
print ("Conjunto a partir de una tupla")
tupla = (1, 2, 3, 4, 5, 5)
conjunto = set(tupla)
print(conjunto)
print(type(conjunto))
```

```python
# Conjunto a partir de una lista
print ("Conjunto a partir de una lista")
lista = [True, False, 0, 1]
conjunto = set(lista)
print(conjunto)
print(type(conjunto))
```

```python
# Conjunto por comprensión
print ("Conjunto por comprensión")
conjunto = {x for x in '🍕🍔🍟🍕🍔🍟🍔🍟'}
print(conjunto)
print(type(conjunto))
```

## Métodos de los conjuntos

Los conjuntos tienen métodos de adición, eliminación, operaciones, búsqueda y copia.

```python
# Método add()
print ("Método add()")
conjunto = {'🍕','🍔','🍟','🌭'}
print(conjunto)
conjunto.add('🥗')
print(conjunto)
```

```python
# Método update()
print ("Método update()")
conjunto = {'🍕','🍔','🍟','🌭'}
print(conjunto)
conjunto.update(['🥤','🍦'])
print(conjunto) 
conjunto.update('🍩🍪')
print(conjunto) 
conjunto.update(('🍫','🍬'))
print(conjunto)
conjunto.update({'🍭','🍮'})
print(conjunto)
```

```python
# Métodos de eliminación: remove(), discard(), pop(), clear()

# remove()
print ("Método remove()")
conjunto = {'🍕','🍔','🍟','🌭'}
print(conjunto)
conjunto.remove('🍔')
print(conjunto)
# conjunto.remove('🍔') # Genera KeyError
```

```python
# discard()
print ("Método discard()")
conjunto = {'🍕','🍔','🍟','🌭'}
print(conjunto)
conjunto.discard('🍔')
print(conjunto)
conjunto.discard('🍔')  # No genera error
print(conjunto)
```

```python
# pop()
print ("Método pop()")
conjunto = {'🍕','🍔','🍟','🌭', '🥤','🍦'}
print(conjunto)
print(conjunto.pop())
print(conjunto)
```

```python
# clear()
print ("Método clear()")
conjunto = {'🍕','🍔','🍟','🌭'}
print(conjunto)
conjunto.clear()
print(conjunto)
```

## Operaciones con conjuntos

- `union()`: Unión de conjuntos  
- `intersection()`: Intersección  
- `difference()`: Diferencia  
- `symmetric_difference()`: Diferencia simétrica


```python
# Unión
print ("Método union()")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
print(c1.union(c2))
```

```python
# Intersección
print ("Método intersection()")
print(c1.intersection(c2))
```

```python
# Diferencia
print ("Método difference()")
print(c1.difference(c2))
print(c2.difference(c1))
```

```python
# Diferencia simétrica
print ("Método symmetric_difference()")
print(c1.symmetric_difference(c2))
```

## Operaciones con asignación: intersection_update(), difference_update(), symmetric_difference_update()


```python
# intersection_update()
print ("Método intersection_update()")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1.intersection_update(c2)
print(c1)
```

```python
# difference_update()
print ("Método difference_update()")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1.difference_update(c2)
print(c1)
```

```python
# symmetric_difference_update()
print ("Método symmetric_difference_update()")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1.symmetric_difference_update(c2)
print(c1)
```

## Métodos de búsqueda: issubset(), issuperset(), isdisjoint()


```python
print ("Método issubset()")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c3 = {'🍔','🍟'}
print(c1.issubset(c2))
print(c3.issubset(c1))
```

```python
print ("Método issuperset()")
print(c1.issuperset(c2))
print(c1.issuperset(c3))
```

```python
print ("Método isdisjoint()")
print(c1.isdisjoint(c2))
print(c1.isdisjoint(c3))
```

## Métodos de copia: copy() y asignación por referencia


```python
print ("Asignación por referencia")
conjunto = {'🍕','🍔','🍟','🌭'}
copia = conjunto
copia.add('🥗')
print(conjunto)
print(copia)
```

```python
print ("Método copy()")
conjunto = {'🍕','🍔','🍟','🌭'}
copia = conjunto.copy()
copia.add('🥗')
print(conjunto)
print(copia)
```

## Funciones incorporadas con conjuntos: len(), max(), min(), sum()


```python
print ("Función len()")
conjunto = {'🍕','🍔','🍟','🌭'}
print(len(conjunto))
```

```python
print ("Función max()")
conjunto = {1, 2, 3, 4, 5}
print(max(conjunto))
conjunto = {'🍕','🍔','🍟','🌭'}
print(max(conjunto))
```

```python
print ("Función min()")
print(min({1, 2, 3, 4, 5}))
print(min({'🍨','🍔','🍟','🍕'}))
```

```python
print ("Función sum()")
print(sum({1, 2, 3, 4, 5}))
```

## Operadores para conjuntos

- `|` unión
- `&` intersección
- `-` diferencia
- `^` diferencia simétrica
- `|=`, `&=`, `-=`, `^=` asignación
- Comparación: `==`, `!=`, `<`, `>`, `<=`, `>=`


```python
print ("Operador |")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
print(c1 | c2)
```

```python
print ("Operador &")
print(c1 & c2)
```

```python
print ("Operador -")
print(c1 - c2)
print(c2 - c1)
```

```python
print ("Operador ^")
print(c1 ^ c2)
```

```python
print ("Operador |= Unión")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1 |= c2
print(c1)
```

```python
print ("Operador &= Intersección")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1 &= c2
print(c1)
```

```python
print ("Operador -= Diferencia")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1 -= c2
print(c1)
```

```python
print ("Operador ^= Diferencia simétrica")
c1 = {'🍔','🍟','🥤'}
c2 = {'🍕','🍨','🥤'}
c1 ^= c2
print(c1)
```

## Conjuntos inmutables: `frozenset()`

Los conjuntos inmutables no pueden modificarse.  
No admiten métodos de adición ni eliminación.


```python
conjunto = frozenset({'🍔','🍕','🥗','🍟','🌭'})
print(conjunto)
print(type(conjunto))
```

## Conjuntos anidados

Se pueden anidar `frozenset` dentro de `set` pero no `set` dentro de `set`.


```python
print ("Conjunto de conjuntos con frozenset")
conjunto = {frozenset({'🍅','🍓','🍎'}), frozenset({'🍈','🍐','🍏'})}
print(conjunto)
```
